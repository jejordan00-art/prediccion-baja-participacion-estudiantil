### Identificación y tratamiento de valores atípicos

Se utiliza el rango intercuartílico (IQR) para identificar observaciones extremas. Un valor atípico estadístico no es necesariamente un error, por lo que no se eliminan registros automáticamente.

**Resultado de la ejecución guardada con los 395 registros de Matemáticas:**

- **Ausencias (`absences_mat`):** el IQR identificó 15 observaciones, con valores entre 21 y 75 ausencias. Los límites calculados fueron -12 y 20. Estos registros están dentro del rango documentado de la variable (0–93), por lo que se conservan y se marcan con `marca_atipico_ausencias`. La decisión es no eliminarlos porque pueden representar ausencias reales.
- **Calificación final (`G3_mat`):** el IQR no identificó observaciones atípicas (0 casos). Los límites calculados fueron -1 y 23; las notas observadas están dentro del rango válido de 0 a 20. Por tanto, no hay registros de G3 que requieran tratamiento por IQR.

La celda siguiente calcula nuevamente los límites, muestra los registros identificados y crea las columnas booleanas de marcado. Si se vuelve a ejecutar con un archivo diferente, deben utilizarse los nuevos resultados impresos.

```python
def detectar_atipicos(data, columna):
    q1 = data[columna].quantile(0.25)
    q3 = data[columna].quantile(0.75)
    iqr = q3 - q1
    limite_inferior = q1 - 1.5 * iqr
    limite_superior = q3 + 1.5 * iqr

    mascara = (data[columna] < limite_inferior) | (data[columna] > limite_superior)
    atipicos = data.loc[mascara, [columna]].copy()
    atipicos.insert(0, "indice_original", atipicos.index)

    print(f"\nVariable: {columna}")
    print(f"Q1: {q1:.2f} | Q3: {q3:.2f} | IQR: {iqr:.2f}")
    print(f"Límite inferior: {limite_inferior:.2f}")
    print(f"Límite superior: {limite_superior:.2f}")
    print(f"Cantidad de valores atípicos: {len(atipicos)}")
    display(atipicos)
    return atipicos

atipicos_ausencias = detectar_atipicos(datos, "absences_mat")
atipicos_calificaciones = detectar_atipicos(datos, "G3_mat")

# Se marcan en la base sin borrar ni modificar el valor original.
datos["marca_atipico_ausencias"] = datos.index.isin(atipicos_ausencias["indice_original"])
datos["marca_atipico_G3"] = datos.index.isin(atipicos_calificaciones["indice_original"])

print("\nRegistros marcados por ausencias:", int(datos["marca_atipico_ausencias"].sum()))
print("Registros marcados por G3:", int(datos["marca_atipico_G3"].sum()))
```

Ingeniería de características y pipeline  

Se crean tres variables derivadas con sentido para el seguimiento académico:

1. `grade_progress_mat`: diferencia entre la segunda y la primera evaluación (`G2 - G1`). Solo es un predictor válido si el seguimiento se realiza después de conocer G2.
2. `has_failures_mat`: indicador de al menos una reprobación previa.
3. `absence_level_mat`: clasificación de ausencias en cuatro niveles.

La función codifica todas las columnas categóricas y escala las ausencias con Min-Max. Las marcas de atípicos se conservan en la base de calidad, pero se excluyen de las variables predictoras. La nota final `G3_mat` se mantiene en la tabla transformada para construir el objetivo, pero se elimina de `X` para evitar fuga de información.

```python
def preparar_features(data):
    """Deriva variables, codifica categorías y escala ausencias de forma reproducible."""
    data = data.copy()

    # 1. Progreso académico entre la primera y segunda evaluación
    data["grade_progress_mat"] = data["G2_mat"] - data["G1_mat"]

    # 2. Indicador de al menos una reprobación previa
    data["has_failures_mat"] = (data["failures_mat"] > 0).astype(int)

    # 3. Clasificación del nivel de ausencias
    data["absence_level_mat"] = pd.cut(
        data["absences_mat"],
        bins=[-1, 0, 5, 15, float("inf")],
        labels=["Sin ausencias", "Bajas", "Medias", "Altas"]
    )

    # Evitar que los indicadores de revisión de calidad entren al modelo
    data = data.drop(
        columns=["marca_atipico_ausencias", "marca_atipico_G3"],
        errors="ignore"
    )

    # Codificar todas las variables categóricas
    columnas_categoricas = data.select_dtypes(
        include=["object", "category"]
    ).columns.tolist()
    if columnas_categoricas:
        data = pd.get_dummies(
            data,
            columns=columnas_categoricas,
            dtype=int
        )

    # Escalado Min-Max de ausencias
    minimo = data["absences_mat"].min()
    maximo = data["absences_mat"].max()
    if maximo != minimo:
        data["absences_scaled"] = (
            (data["absences_mat"] - minimo) / (maximo - minimo)
        )
    else:
        data["absences_scaled"] = 0.0

    return data

# Ejecutar el pipeline
UMBRAL_BAJO_RENDIMIENTO = 10
df_features = preparar_features(datos)
print("Dimensiones de la base transformada:", df_features.shape)
display(df_features.head())

# Verificar que no queden columnas object/category sin codificar
columnas_categoricas_pendientes = df_features.select_dtypes(
    include=["object", "category"]
).columns.tolist()
print("Columnas categóricas pendientes:", columnas_categoricas_pendientes)

# Comprobar reproducibilidad
# Dos ejecuciones sobre la misma entrada deben producir exactamente la misma salida.
df_v1 = preparar_features(datos)
df_v2 = preparar_features(datos)
print("¿El pipeline es reproducible?", df_v1.equals(df_v2))
```

## Segmentación y asociación

### Segmentación inicial con K-means

Se aplica K-means sobre variables disponibles antes del cierre del periodo: edad, tiempo de estudio, reprobaciones previas, ausencias y calificaciones de los dos primeros periodos (G1 y G2). La calificación final (`G3_mat`) no participa en el entrenamiento de los grupos; se utiliza únicamente después para describir los segmentos.

Las variables se estandarizan antes de aplicar K-means. Además de comparar el coeficiente Silhouette, se incorpora el método del codo solicitado en la guía: se calcula la inercia para valores de k entre 1 y 6. La selección de k es exploratoria y debe justificarse observando la curva, la separación de los grupos y su interpretación.

```python
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

columnas_segmentacion = [
    "age_mat", "studytime_mat", "failures_mat",
    "absences_mat", "G1_mat", "G2_mat"
]

# Se utiliza la misma matriz numérica y estandarizada para comparar técnicas.
X_segmentacion = datos[columnas_segmentacion].copy()
escalador_segmentacion = StandardScaler()
X_segmentacion_escalado = escalador_segmentacion.fit_transform(X_segmentacion)

# 1. Método del codo: calcular inercia para k de 1 a 6.
resultados_codo = []
for k in range(1, 7):
    modelo_k = KMeans(n_clusters=k, random_state=42, n_init=10)
    modelo_k.fit(X_segmentacion_escalado)
    resultados_codo.append({"k": k, "inercia": modelo_k.inertia_})

tabla_codo = pd.DataFrame(resultados_codo)
print("Método del codo (buscar dónde la inercia deja de disminuir fuertemente):")
display(tabla_codo.round(3))

plt.figure(figsize=(7, 4))
plt.plot(tabla_codo["k"], tabla_codo["inercia"], marker="o")
plt.xlabel("Número de grupos (k)")
plt.ylabel("Inercia")
plt.title("Método del codo para K-means")
plt.xticks(range(1, 7))
plt.grid(alpha=0.3)
plt.show()

# 2. Comparar también Silhouette para k entre 2 y 5.
resultados_silhouette = []
for k in range(2, 6):
    modelo_k = KMeans(n_clusters=k, random_state=42, n_init=10)
    etiquetas_k = modelo_k.fit_predict(X_segmentacion_escalado)
    resultados_silhouette.append({
        "k": k,
        "silhouette": silhouette_score(X_segmentacion_escalado, etiquetas_k)
    })

comparacion_k = pd.DataFrame(resultados_silhouette).sort_values(
    "silhouette", ascending=False
)
print("Comparación complementaria con Silhouette:")
display(comparacion_k.round(3))

# Selección provisional. Revisa la curva del codo y cambia este valor si tu análisis lo justifica.
# Se deja 5 para mantener la interpretación de cinco grupos que ya estaba en el notebook.
k_elegido = 5
print("Número de grupos utilizado en esta ejecución:", k_elegido)

modelo_segmentacion = KMeans(
    n_clusters=k_elegido, random_state=42, n_init=10
)
clusters = modelo_segmentacion.fit_predict(X_segmentacion_escalado)

resultado_segmentacion = datos.copy()
resultado_segmentacion["cluster"] = clusters
resultado_segmentacion["bajo_rendimiento"] = y.values

resumen_clusters = resultado_segmentacion.groupby("cluster").agg(
    estudiantes=("G3_mat", "size"),
    promedio_G3=("G3_mat", "mean"),
    promedio_ausencias=("absences_mat", "mean"),
    promedio_tiempo_estudio=("studytime_mat", "mean"),
    promedio_reprobaciones=("failures_mat", "mean"),
    porcentaje_bajo_rendimiento=(
        "bajo_rendimiento", lambda s: s.mean() * 100
    )
).round(2)

print("Perfil descriptivo de los segmentos K-means:")
display(resumen_clusters)
```

Los números de cluster son etiquetas técnicas asignadas por K-means y no representan un orden. Los perfiles describen asociaciones dentro de esta muestra; no demuestran causas. La elección final de K debe validarse con las instrucciones de la sesión en vivo.

Conclusión de la segmentación:

Eso sí, este ejercicio es puramente exploratorio y no implica causalidad. La guía de Unidad 5 pide aplicar clustering jerárquico sobre la misma matriz estandarizada `X_segmentacion_escalado` y comparar las etiquetas con las de K-means mediante una tabla cruzada. Se utiliza el mismo número de grupos elegido para K-means y el método `ward`.

### Segmentación con DBSCAN

DBSCAN agrupa observaciones por densidad y puede marcar registros aislados con la etiqueta `-1` (ruido u observaciones atípicas). Como el resultado depende de `eps`, se prueban varios valores con `min_samples=5`. Se busca una solución con al menos dos grupos y sin que todos los registros queden en un solo grupo o marcados como ruido.

La siguiente celda compara los parámetros candidatos. Si ninguno produce una solución interpretable, amplía o ajusta la lista `eps_candidatos` y vuelve a ejecutar.

```python
from sklearn.cluster import DBSCAN

# Valores candidatos de eps para la matriz estandarizada de seis variables.
eps_candidatos = [0.8, 1.0, 1.2, 1.5, 1.8, 2.0, 2.5, 3.0]
min_samples_dbscan = 5
resultados_dbscan = []

for eps in eps_candidatos:
    etiquetas = DBSCAN(
        eps=eps,
        min_samples=min_samples_dbscan
    ).fit_predict(X_segmentacion_escalado)

    grupos = set(etiquetas) - {-1}
    cantidad_grupos = len(grupos)
    cantidad_ruido = int((etiquetas == -1).sum())
    porcentaje_ruido = cantidad_ruido / len(etiquetas) * 100

    # Silhouette se calcula solo sobre puntos que DBSCAN asignó a un grupo,
    # y únicamente si hay al menos dos grupos válidos.
    mascara_no_ruido = etiquetas != -1
    silhouette_db = np.nan
    etiquetas_sin_ruido = etiquetas[mascara_no_ruido]
    if (cantidad_grupos >= 2 and
        len(set(etiquetas_sin_ruido)) >= 2 and
        len(set(etiquetas_sin_ruido)) < len(etiquetas_sin_ruido)):
        silhouette_db = silhouette_score(
            X_segmentacion_escalado[mascara_no_ruido],
            etiquetas_sin_ruido
        )

    resultados_dbscan.append({
        "eps": eps,
        "cantidad_grupos": cantidad_grupos,
        "registros_ruido": cantidad_ruido,
        "porcentaje_ruido": round(porcentaje_ruido, 2),
        "silhouette_sin_ruido": silhouette_db
    })

tabla_dbscan = pd.DataFrame(resultados_dbscan)
print("Comparación de parámetros para DBSCAN (-1 representa ruido):")
display(tabla_dbscan.round(3))

# Criterio auxiliar para recomendar eps: al menos 2 grupos, Silhouette calculable
# y no más del 50% de registros marcados como ruido. El 50% es una referencia
# práctica para esta exploración, no una regla universal del algoritmo.
soluciones_validas = tabla_dbscan[
    (tabla_dbscan["cantidad_grupos"] >= 2) &
    (tabla_dbscan["porcentaje_ruido"] <= 50) &
    (tabla_dbscan["silhouette_sin_ruido"].notna())
]

if not soluciones_validas.empty:
    fila_elegida = soluciones_validas.sort_values(
        "silhouette_sin_ruido", ascending=False
    ).iloc[0]
    eps_elegido = float(fila_elegida["eps"])
    modelo_dbscan = DBSCAN(
        eps=eps_elegido,
        min_samples=min_samples_dbscan
    )
    etiquetas_dbscan = modelo_dbscan.fit_predict(X_segmentacion_escalado)
    resultado_segmentacion["cluster_dbscan"] = etiquetas_dbscan

    print(f"Valor eps seleccionado de forma preliminar: {eps_elegido}")
    print("Cantidad de grupos (sin contar -1):", len(set(etiquetas_dbscan) - {-1}))
    print("Distribución de etiquetas DBSCAN (-1 = ruido):")
    display(pd.Series(etiquetas_dbscan, name="cluster_dbscan").value_counts().sort_index())
else:
    print(
        "No se encontró un eps candidato con al menos dos grupos, Silhouette calculable "
        "y como máximo 50% de ruido. Ajusta eps_candidatos o min_samples y vuelve a ejecutar. "
        "No reportes DBSCAN como una segmentación útil hasta obtener una solución interpretable."
    )
```

## Modelo predictivo y evaluación

Este contenido, junto con la segmentación de la Unidad 5, forma parte del avance del Checkpoint 2 (20%) indicado en las guías de clase.

### Definición del objetivo y separación de datos

Se construye un modelo de **clasificación** para identificar estudiantes con bajo rendimiento al finalizar el periodo. El objetivo es `bajo_rendimiento`: vale 1 cuando `G3_mat < 10` y 0 cuando `G3_mat >= 10`.

Se utiliza `X` como matriz de predictores y `y` como variable objetivo, definidas en la sección anterior. `G3_mat` debe permanecer excluida de `X`, porque es la nota que se busca predecir. Los predictores `G1_mat`, `G2_mat` y `grade_progress_mat` solo son válidos si la alerta se realiza después de conocer la segunda evaluación, tal como se indicó en el alcance temporal del proyecto.

```python
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import (
    accuracy_score,
    confusion_matrix,
    precision_score,
    recall_score,
    f1_score
)

# Comprobación de fuga directa: la nota final y la variable objetivo no pueden estar en X.
columnas_prohibidas = ["G3_mat", "bajo_rendimiento"]
columnas_prohibidas_presentes = [
    columna for columna in columnas_prohibidas if columna in X.columns
]
print("Columnas de resultado que aparecen en X:", columnas_prohibidas_presentes)
assert not columnas_prohibidas_presentes, "Hay fuga directa: elimina la variable objetivo de X."

# Separar entrenamiento (70 %) y prueba (30 %), conservando la proporción de clases.
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.30,
    random_state=42,
    stratify=y
)

print("Tamaño de entrenamiento:", X_train.shape)
print("Tamaño de prueba:", X_test.shape)

# Primer modelo: árbol de decisión con profundidad máxima 4.
modelo_arbol = DecisionTreeClassifier(
    max_depth=4,
    random_state=42
)
modelo_arbol.fit(X_train, y_train)

pred_train = modelo_arbol.predict(X_train)
pred_test = modelo_arbol.predict(X_test)

accuracy_train = accuracy_score(y_train, pred_train)
accuracy_test = accuracy_score(y_test, pred_test)

print("Accuracy de entrenamiento:", round(accuracy_train, 3))
print("Accuracy de prueba:", round(accuracy_test, 3))

# Métricas de evaluación para la clase positiva 1 = bajo rendimiento.
matriz = confusion_matrix(y_test, pred_test, labels=[0, 1])
print("Matriz de confusión (filas = valor real; columnas = predicción):")
display(pd.DataFrame(
    matriz,
    index=["Real: aprobación", "Real: bajo rendimiento"],
    columns=["Predicho: aprobación", "Predicho: bajo rendimiento"]
))

precision = precision_score(y_test, pred_test, pos_label=1, zero_division=0)
recall = recall_score(y_test, pred_test, pos_label=1, zero_division=0)
f1 = f1_score(y_test, pred_test, pos_label=1, zero_division=0)

print("Precisión para bajo rendimiento:", round(precision, 3))
print("Exhaustividad (recall) para bajo rendimiento:", round(recall, 3))
print("F1 para bajo rendimiento:", round(f1, 3))

# Referencia simple: predecir siempre la clase más frecuente del entrenamiento.
clase_mayoritaria = int(y_train.mode().iloc[0])
pred_baseline = np.full(len(y_test), clase_mayoritaria)
accuracy_baseline = accuracy_score(y_test, pred_baseline)
print("Accuracy de referencia (predecir siempre la clase mayoritaria):", round(accuracy_baseline, 3))

# Variables que más utilizó el árbol.
importancias = pd.Series(
    modelo_arbol.feature_importances_,
    index=X.columns
).sort_values(ascending=False)
print("15 variables más importantes según el árbol:")
display(importancias.head(15).to_frame("importancia"))
```

### Revisión de fuga de datos y sobreajuste

La comprobación de fuga directa debe mostrar una lista vacía y `G3_mat` debe estar ausente de `X`. La variable `grade_progress_mat` se calcula con `G1_mat` y `G2_mat`, no con `G3_mat`; por eso solo es válida bajo el alcance temporal de una alerta generada después de la segunda evaluación.

Para revisar sobreajuste, se compara el accuracy de entrenamiento con el accuracy de prueba. Si el resultado de entrenamiento es mucho mayor que el de prueba, puede indicar que el árbol memorizó patrones específicos. No existe un único umbral universal: interpreta la diferencia junto con la matriz de confusión, precisión, recall y F1.

```python
brecha_accuracy = accuracy_train - accuracy_test
print("Diferencia entre accuracy de entrenamiento y prueba:", round(brecha_accuracy, 3))

if brecha_accuracy > 0.10:
    print("Advertencia: la diferencia es amplia y puede indicar sobreajuste. Revisa la complejidad del árbol.")
else:
    print("No se observa una brecha superior a 0.10 en esta ejecución. Aun así, revisa todas las métricas y la matriz de confusión.")
```

### Reto opcional de Unidad 6: comparar distintas profundidades del árbol

Se elige la opción 2 de la guía: probar profundidades 2, 4, 8 y sin límite (`None`), comparando accuracy de entrenamiento y prueba. Si la precisión de entrenamiento aumenta mucho y la de prueba se estanca o disminuye, puede ser señal de sobreajuste.

```python
resultados_profundidad = []

for profundidad in [2, 4, 8, None]:
    arbol = DecisionTreeClassifier(
        max_depth=profundidad,
        random_state=42
    )
    arbol.fit(X_train, y_train)
    pred_train_profundidad = arbol.predict(X_train)
    pred_test_profundidad = arbol.predict(X_test)

    resultados_profundidad.append({
        "max_depth": "Sin límite" if profundidad is None else profundidad,
        "accuracy_entrenamiento": accuracy_score(y_train, pred_train_profundidad),
        "accuracy_prueba": accuracy_score(y_test, pred_test_profundidad)
    })

tabla_profundidades = pd.DataFrame(resultados_profundidad)
tabla_profundidades["brecha"] = (
    tabla_profundidades["accuracy_entrenamiento"] -
    tabla_profundidades["accuracy_prueba"]
)
print("Comparación de profundidades del árbol:")
display(tabla_profundidades.round(3))
```
